## Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

##Upload an Image

```python
from google.colab import files

uploaded = files.upload()
```

## Get the Uploaded Filename

```python
filename = list(uploaded.keys())[0]

print("Uploaded file:", filename)
```

## Read the Image

```python
img = cv2.imread(filename)

if img is None:
    print("Error: Image could not be loaded.")
else:
    print("Image loaded successfully!")
```

## Display the Original Image

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

## Display Image Information

```python
height, width, channels = img.shape

print("Image Information")
print("------------------")
print("Height:", height, "pixels")
print("Width :", width, "pixels")
print("Channels:", channels)
print("Total pixels:", height * width)
```

## Convert to Grayscale

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print("Grayscale conversion completed!")
```

## Display Grayscale Image

```python
plt.figure(figsize=(8, 6))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

## Check Grayscale Image Properties

```python
print("Original image shape :", img.shape)
print("Grayscale image shape:", gray.shape)
```

## Resize the Image

```python
resized = cv2.resize(img, (400, 300))

print("Resized image shape:", resized.shape)
```

##  Display Resized Image

```python
resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(resized_rgb)
plt.title("Resized Image - 400 x 300")
plt.axis("off")
plt.show()
```

## Compare Original and Grayscale

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

## Save the Grayscale Image

```python
gray_filename = "gray_image.jpg"

cv2.imwrite(gray_filename, gray)

print("Grayscale image saved as:", gray_filename)
```

## Download the Processed Image

```python
from google.colab import files

files.download(gray_filename)
```

## Basic Pixel Information

```python
print("Pixel value at (0, 0):", img[0, 0])
print("Grayscale pixel value at (0, 0):", gray[0, 0])
```

## Display a Small Image Region

```python
small_region = img_rgb[0:200, 0:200]

plt.figure(figsize=(5, 5))
plt.imshow(small_region)
plt.title("Top-Left 200 x 200 Region")
plt.axis("off")
plt.show()
```

## Modify a Small Region

```python
modified = img.copy()

modified[0:150, 0:150] = [0, 0, 0]

modified_rgb = cv2.cvtColor(modified, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(modified_rgb)
plt.title("Image with Modified Region")
plt.axis("off")
plt.show()
```

## Save the Modified Image

```python
modified_filename = "modified_image.jpg"

cv2.imwrite(modified_filename, modified)

print("Modified image saved as:", modified_filename)
```

## Download Modified Image

```python
files.download(modified_filename)
```

## Final Comparison

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(modified_rgb)
plt.title("Modified")
plt.axis("off")

plt.show()
```
